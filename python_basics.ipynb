## 3. Strings and String Methods

Strings are sequences of characters. They are used for names, messages, addresses, and other text data.

```python
# String creation and formatting
first_name = "Taylor"
last_name = "Jordan"
full_name = first_name + " " + last_name

print("Hello, " + full_name)
print(f"Welcome, {full_name}!")

# Real-life example: build notification messages or customer greetings.
```

### String slicing and indexing

We can extract parts of text and inspect individual characters.

### Common string methods

Methods help clean and transform text, such as changing case, removing spaces, or searching.

```python
sample = "  welcome to python programming  "
print("Uppercase:", sample.upper())
print("Lowercase:", sample.lower())
print("Strip spaces:", sample.strip())
print("Replace text:", sample.replace("python", "Python"))
print("Contains 'python'?:", "python" in sample.lower())
```

## 4. Lists and List Operations

Lists hold multiple values in order. They are great for shopping lists, task lists, or collecting data points.

```python
# Create a list of shopping items
shopping_list = ["milk", "eggs", "bread", "apples"]
print(shopping_list)

# Access items by index
print("First item:", shopping_list[0])
print("Last item:", shopping_list[-1])

# Add an item and display length
shopping_list.append("cheese")
print("Updated list:", shopping_list)
print("Total items:", len(shopping_list))
```

### List methods

Python lists come with built-in methods for adding, removing, sorting, and checking items.

```python
fruits = ["apple", "banana", "cherry"]
print("Original list:", fruits)
fruits.append("date")
print("After append:", fruits)
fruits.insert(1, "blueberry")
print("After insert:", fruits)
fruits.remove("banana")
print("After remove:", fruits)
print("Index of cherry:", fruits.index("cherry"))
fruits.sort()
print("Sorted list:", fruits)
print("Count of apple:", fruits.count("apple"))
```

### List slicing and iteration

Lists can be sliced and looped through to process each item.

```python
numbers = [2, 4, 6, 8, 10]
print("First three:", numbers[:3])
print("Every other item:", numbers[::2])

for number in numbers:
    print("Number:", number)

# Real-life example: summing sales amounts or printing a list of tasks.
```

## 5. Tuples and When to Use Them

Tuples are like lists but cannot be changed after creation. Use tuples for fixed groups of values.

```python
# Create a tuple
coordinate = (37.7749, -122.4194)
print("Coordinate:", coordinate)

# Unpack values
latitude, longitude = coordinate
print("Latitude:", latitude)
print("Longitude:", longitude)

# Real-life use case: store fixed configuration values, GPS coordinates, or dates.
```

### Tuple immutability

Trying to change a tuple item causes an error, which is useful when values should remain constant.

```python
immutable_tuple = (1, 2, 3)
print("Tuple values:", immutable_tuple)

# immutable_tuple[0] = 5  # This would raise an error

# Convert tuple to list if you need to modify it
mutable_copy = list(immutable_tuple)
mutable_copy.append(4)
print("Modified copy:", mutable_copy)
```

## 6. Dictionaries and Key-Value Mapping

Dictionaries store values by key. They are useful for mapping names to data, like user profiles or product prices.

```python
# Create a dictionary for a student record
student = {
    "name": "Priya",
    "age": 17,
    "grade": "A",
    "passed": True
}

print("Student name:", student["name"])
print("Grade:", student.get("grade"))

# Update a value
student["grade"] = "A+"
print("Updated record:", student)
```

### Dictionary methods

Dictionaries include methods for working with keys, values, and items in mappings.

```python
product = {"name": "Notebook", "price": 4.5, "stock": 12}
print("Original product:", product)
print("Keys:", list(product.keys()))
print("Values:", list(product.values()))
print("Items:", list(product.items()))
product.update({"price": 4.99, "color": "blue"})
print("After update:", product)
print("Has stock key?:", "stock" in product)
print("Price with get():", product.get("price"))
removed = product.pop("stock")
print("Removed stock:", removed)
print("Final product:", product)
```

### Dictionary iteration

Loop through keys, values, or pairs to process dictionary data.

```python
inventory = {
    "pen": 12,
    "notebook": 5,
    "eraser": 3
}

for item, count in inventory.items():
    print(f"{item}: {count}")

# Real-life example: manage stock levels or smartphone contacts.
```

## 7. Control Flow: if, elif, else

Conditionals let your program make decisions, such as checking age or calculating grades.

```python
age = 15
if age < 13:
    print("Child fare")
elif age < 18:
    print("Teen fare")
else:
    print("Adult fare")

# Real-life example: membership discounts, login checks, or age verification.
```

### Grading example

Use `if`, `elif`, and `else` to convert numeric grades to letter grades.

```python
score = 82
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score: {score}, Grade: {grade}")
```

## 8. Loops: for and while

Loops repeat code for each item or until a condition changes.

```python
# for loop with a list
colors = ["red", "green", "blue"]
for color in colors:
    print("Color:", color)

# while loop with a counter
countdown = 5
while countdown > 0:
    print("Countdown:", countdown)
    countdown -= 1
print("Blast off!")

# Real-life example: process user inputs, generate reports, or count events.
```

### Loop with conditional logic

Combine loops and conditionals to filter or categorize values.

```python
numbers = [3, 10, 7, 2, 11, 8]
for num in numbers:
    if num % 2 == 0:
        print(num, "is even")
    else:
        print(num, "is odd")
```

## 9. Functions and Code Reuse

Functions package reusable logic. They make code easier to read and maintain.

```python
def greet_user(name):
    return f"Hello, {name}! Welcome to Python."

message = greet_user("Sonia")
print(message)

# Function with multiple parameters

def calculate_area(width, height):
    return width * height

area = calculate_area(5, 10)
print("Area:", area)

# Real-life example: reuse calculations, text formatting, or validation logic.
```

### Functions with default values and docstrings

Default arguments let functions work with common cases automatically.

```python
def make_greeting(name, language="English"):
    if language == "Spanish":
        return f"Hola, {name}!"
    elif language == "French":
        return f"Bonjour, {name}!"
    return f"Hello, {name}!"

print(make_greeting("Mina"))
print(make_greeting("Mina", language="Spanish"))
```
